```python
import sqlite3
import pandas as pd

DB_PATH  = "data/option_chain.db"
CSV_PATH = "data/option_chain.csv"

with sqlite3.connect(DB_PATH) as conn:
    df = pd.read_sql("SELECT * FROM nifty50_option_chain ORDER BY timestamp", conn)

print(f"Total rows: {len(df)}")
df.head()
```

```python
df.to_csv(CSV_PATH, index=False)
print(f"Saved to {CSV_PATH}")
```
